```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from sklearn.linear_model import Lasso
from pysindy import SINDy
from pykoopman import Koopman
from pykoopman.observables import PolynomialObservables

# Define the nonlinear system
def nonlinear_system(t, x):
    return -x + x**3

# Generate synthetic data
t_span = (0, 10)
t_eval = np.linspace(t_span[0], t_span[1], 1000)
x0 = [0.5]  # Initial condition
sol = solve_ivp(nonlinear_system, t_span, x0, t_eval=t_eval)
x = sol.y.T
t = sol.t

# Compute time derivatives (for training)
x_dot = np.array([nonlinear_system(None, xi) for xi in x])

# Define observables (polynomials up to degree 3)
observables = PolynomialObservables(degree=3)

# Fit the Koopman operator
koopman = Koopman(observables=observables)
koopman.fit(x, t=t)

# Lift the data into the observable space
g_x = koopman.observables.transform(x)

# Use LASSO to identify the Koopman operator
model = Lasso(alpha=0.01)  # Regularization parameter
model.fit(g_x, x_dot)

# Extract the Koopman operator
K = model.coef_

# Print the identified Koopman operator
print("Identified Koopman operator (K):")
print(K)

# Compare the predicted and true dynamics
x_dot_pred = model.predict(g_x)

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(t, x_dot, label='True $\dot{x}$', linewidth=2)
plt.plot(t, x_dot_pred, '--', label='Predicted $\dot{x}$', linewidth=2)
plt.xlabel('Time')
plt.ylabel('$\dot{x}$')
plt.legend()
plt.title('True vs Predicted Dynamics')
plt.show()
```

```python
import os
import numpy as np

def load_files_with_pattern(folder_path, pattern):
    """
    Load files from a folder into two arrays based on a pattern in the filename.

    Args:
        folder_path (str): Path to the folder containing the files.
        pattern (str): Pattern to match in filenames.

    Returns:
        matched_data (list): List of arrays loaded from files matching the pattern.
        other_data (list): List of arrays loaded from files not matching the pattern.
    """
    matched_data = []
    other_data = []

    # Iterate over all files in the folder
    for filename in os.listdir(folder_path):
        file_path = os.path.join(folder_path, filename)

        # Check if the file is a valid file (not a directory)
        if os.path.isfile(file_path):
            # Load the file using numpy
            data = np.load(file_path)

            # Check if the filename contains the pattern
            if pattern in filename:
                matched_data.append(data)
            #
            #else:
            #    other_data.append(data)

    return matched_data

# Example usage
folder_path = "data/SR_Mc_good/"  # Replace with your folder path
pattern = "9.57"  # Replace with your desired pattern

matched_data = load_files_with_pattern(folder_path, pattern)

print(f"Number of files matching the pattern: {len(matched_data)}")
#print(f"Number of other files: {len(other_data)}")
```
