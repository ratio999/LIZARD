```python
# The outputs of this function are the "plus" and "cross" polarizations 
# of the gravitational-wave signal as viewed from the line of sight at 
# a given source inclination (assumed face-on, i.e. zero inclination
# if not provided)
import matplotlib.pyplot as pylab
from pycbc.waveform import get_td_waveform, fd_approximants
hp, hc = get_td_waveform(approximant="IMRPhenomD",
                         mass1=10,
                         mass2=10,
                         delta_t=1.0/16384,
                         f_lower=30)

pylab.figure(figsize=pylab.figaspect(0.4))
pylab.plot(hp.sample_times, hp, label='Plus Polarization')
pylab.plot(hp.sample_times, hc, label='Cross Polarization')
pylab.xlabel('Time (s)')
pylab.ylabel('Strain')
pylab.legend()
pylab.grid()
pylab.show()

# Zoom in near the merger time
pylab.figure(figsize=pylab.figaspect(0.4))
pylab.plot(hp.sample_times, hp, label='Plus Polarization')
pylab.plot(hp.sample_times, hc, label='Cross Polarization')
pylab.xlabel('Time (s)')
pylab.ylabel('Strain')
pylab.xlim(-1, .01)
pylab.legend()
pylab.grid()
pylab.show()
```

```python
from scipy.signal import hilbert, savgol_filter
from scipy.interpolate import CubicSpline
# Define the spline regions
spline_length = 1000  # Number of samples to spline at the beginning and end
fs = 16384
# Create a spline function for the beginning
x_begin = np.linspace(0, 1, spline_length)
spline_begin = CubicSpline([0, 1], [0, 1], bc_type=((1, 0), (2, 0)))  # Smooth transition from 0 to 1

# Create a spline function for the end
x_end = np.linspace(0, 1, spline_length)
spline_end = CubicSpline([0, 1], [1, 0], bc_type=((1, 0), (2, 0)))  # Smooth transition from 1 to 0

# Apply the spline to the signal
splined_signal = signal.copy()
splined_signal[:spline_length] *= spline_begin(x_begin)  # Spline the beginning
#splined_signal[len(signal)-spline_length:] *= spline_end(x_end)  # Spline the end

# Compute the analytic signal using the Hilbert transform
analytic_signal = hilbert(splined_signal)

# Compute the instantaneous phase
instantaneous_phase = np.unwrap(np.angle(analytic_signal))

# Compute the instantaneous frequency
instantaneous_frequency = np.diff(instantaneous_phase) / (2.0 * np.pi) * fs

# Smooth the instantaneous frequency using a Savitzky-Golay filter
from scipy.ndimage import gaussian_filter
smoothed_frequency = gaussian_filter(instantaneous_frequency, sigma=2)
#smoothed_frequency = savgol_filter(instantaneous_frequency, window_length=51, polyorder=3)
#smoothed_frequency = instantaneous_frequency
fisco = np.argmax(smoothed_frequency)
smoothed_frequency=smoothed_frequency[fisco-10000:fisco]
# Adjust the time vector to match the instantaneous frequency
t_inst_freq = times[:-1]  # Remove the last time step to match the length of instantaneous_frequency
t_inst_freq = t_inst_freq[fisco-10000:fisco]
# Plot the original signal, splined signal, and instantaneous frequency
pylab.figure(figsize=(12, 8))

# Plot the original signal
pylab.subplot(3, 1, 1)
pylab.plot(times, signal, label="Original Signal", color="blue")
pylab.xlabel("Time (s)")
pylab.ylabel("Amplitude")
pylab.title("Original Signal")
pylab.legend()

# Plot the splined signal
pylab.subplot(3, 1, 2)
pylab.plot(times, splined_signal, label="Splined Signal", color="green")
pylab.xlabel("Time (s)")
pylab.ylabel("Amplitude")
pylab.title("Signal with Splined Edges")
pylab.legend()


dt = t_inst_freq[1] - t_inst_freq[0]  # Step size
first_derivative = np.gradient(smoothed_frequency, dt)
#first_derivative = savgol_filter(first_derivative, window_length=51, polyorder=3)
# Plot the smoothed instantaneous frequency
pylab.subplot(3, 1, 3)
pylab.plot(t_inst_freq, first_derivative, label="Smoothed Instantaneous Frequency", color="red")
pylab.xlabel("Time (s)")
pylab.ylabel("Frequency (Hz)")
pylab.title("Smoothed Instantaneous Frequency (Chirp)")
pylab.legend()

pylab.tight_layout()
pylab.show()
```
