```python
import gwpy
from pycbc.waveform import get_td_waveform
from gwosc.datasets import event_gps
from gwpy.timeseries import TimeSeries

from pycbc.waveform import get_td_waveform, fd_approximants
hp, hc = get_td_waveform(approximant="IMRPhenomD",
                         mass1=15,
                         mass2=15,
                         delta_t=1.0/16384,
                         f_lower=30)

plt.figure(figsize=plt.figaspect(0.4))
plt.plot(hp.sample_times, hp, label='Plus Polarization')
plt.plot(hp.sample_times, hc, label='Cross Polarization')
plt.xlabel('Time (s)')
plt.ylabel('Strain')
plt.legend()
plt.grid()
plt.show()

# Zoom in near the merger time
plt.figure(figsize=plt.figaspect(0.4))
plt.plot(hp.sample_times, hp, label='Plus Polarization')
plt.plot(hp.sample_times, hc, label='Cross Polarization')
plt.xlabel('Time (s)')
plt.ylabel('Strain')
plt.xlim(-1, .01)
plt.legend()
plt.grid()
plt.show()
```

```python
from utils import basic_relations as br
Mc = br.Mc_of_m1_m2(10,10)
t = br.ChirpTime(50, br.Mc_of_m1_m2(10,10))
array = np.asarray(hp.sample_times)
idx = np.abs(array + t).argmin()
#t_ind = np.argmin(abs(t[0] + hp.sample_times))
arg = np.argmax(abs(hp))
#idx, arg = 0, -1
signal, times = hp[idx:arg], hp.sample_times[idx:arg]
renorm = max(np.correlate(signal, signal, mode='full'))**0.5
signal /= renorm
print(signal.shape, times.shape)
plt.plot(times, signal)
plt.show()
```

```python
import pywt
scales = np.arange(1,128)
coeffs, freqs = pywt.cwt(signal, scales, 'shan', sampling_period = times[2]-times[0])
```

```python
plt.figure(figsize=(10,6))
plt.imshow(np.abs(coeffs), extent=[times[0], times[-1], freqs[-1], freqs[0]], cmap='viridis', aspect='auto')
plt.xlabel('Time [s]')
plt.ylabel('Frequency [Hz]')
plt.title("CWT-Morlet")
plt.colorbar(label='Magintude')
plt.show()
print(coeffs[np.where(np.isnan(coeffs))])
print(times.shape, freqs.shape, coeffs.shape)
```

```python
data = np.empty((8446*127, 3))
for i in range (0,8446):
    for j in range (0,127):
        data[i*127+j,:] = times[i], freqs[j], coeffs[j,i]
```
