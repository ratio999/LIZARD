```python
def redistribute_ys_multinomial_corrected(xs, ys, zs, ars, y_intervals=None):
    """
    Redistribuisce i valori ys secondo una distribuzione multinomiale
    basata sui valori zs (posterior_per_area).
    
    Per ogni valore unico di xs:
    1. Trova tutti i punti con quel valore di xs
    2. Usa i loro valori zs come pesi per una multinomiale
    3. Redistribuisce i punti negli intervalli y secondo questi pesi
    4. All'interno di ogni intervallo, usa una distribuzione beta per concentrare
    """

    
    # Identifica i valori unici di x
    unique_xs, counts_xs = np.unique(xs, return_counts=True)
    print(f"Valori unici di xs: {unique_xs}")
    
    # Identifica i valori unici di y (centri degli intervalli)
    
    #print(f"Valori unici di ys (centri intervalli): {unique_ys}")
    
    # Se non sono specificati gli intervalli, li calcoliamo
    
    xs_redistributed = []
    ys_redistributed = []
    zs_redistributed = []
    ars_redistributed = []
    #zs *= ars
    # Per ogni valore unico di x
    for x_val in unique_xs:
        # Trova tutti i punti con questo valore di x
        indices = np.where(xs == x_val)[0]
        n_points = len(indices)
        yx = ys[indices]
        z_values = zs[indices]
        
        if np.sum(z_values)>= 0*np.sqrt(np.sum(zs)/len(xs)):
            unique_ys, counts_ys = np.unique(yx, return_counts=True)
            y_intervals = []
            print('YS: ', unique_ys)
            for i, y_val in enumerate(unique_ys):
                if i == 0:
                    if len(unique_ys) > 1:
                        y_next = unique_ys[i+1]
                        #left_bound = y_val - (y_next - y_val) / 2
                        right_bound = (y_val + y_next) / 2
                        left_bound = 2*y_val - right_bound
                    else:
                        left_bound = y_val - 1e-10
                        right_bound = y_val + - 1e-10
                elif i == len(unique_ys) - 1:
                    y_prev = unique_ys[i-1]
                    left_bound = (y_prev + y_val) / 2
                    right_bound = y_val + (y_val - y_prev) / 2
                else:
                    y_prev = unique_ys[i-1]
                    y_next = unique_ys[i+1]
                    left_bound = (y_prev + y_val) / 2
                    right_bound = (y_val + y_next) / 2

                y_intervals.append((left_bound, right_bound))
            print(f"\Processando x = {x_val} con {n_points} punti")

            # Estrai i valori z (posterior) per questi punti

            print(f"  Valori z (posterior): {z_values}")

            # Normalizza i valori z per ottenere le probabilità della multinomiale
            z_probs = z_values / np.sum(z_values)
            print(f"  Probabilità normalizzate: {z_probs}")

            # Usa la multinomiale per decidere quanti punti vanno in ogni intervallo
            # np.random.multinomial(n_trials, probabilities) restituisce il numero di successi per ogni categoria
            interval_counts = np.random.multinomial(n_points, z_probs)
            print(f"  Punti assegnati per intervallo: {interval_counts}")

            # Redistribuisci i punti negli intervalli
            for i, count in enumerate(interval_counts):
                if count > 0:
                    left_bound, right_bound = y_intervals[i]

                    # Usa il valore z corrispondente per determinare la concentrazione beta
                    z_val = z_values[i]
                    if z_val >= np.sqrt(np.sum(zs)/len(xs)/len(indices)):  
                        # Parametri beta: più alto z_val = più concentrazione verso il centro
                        # Mappiamo z_val a parametri alpha e beta
                        # Se z_val è alto, vogliamo alta concentrazione (alpha=beta alti)
                        # Se z_val è basso, vogliamo distribuzione più uniforme (alpha=beta bassi)

                        # Normalizziamo z_val rispetto al range dei valori z
                        '''z_min, z_max = np.min(z_values), np.max(z_values)
                        if z_max > z_min:
                            z_normalized = z_val
                        else:
                            z_normalized = z_min'''
                        z_normalized = 1
                        # Mappiamo a parametri beta: [1,1] uniforme -> [10,10] molto concentrato
                        alpha = beta_param = z_normalized  # Range [1, 10]

                        print(f"    Intervallo {i}: [{left_bound:.3f}, {right_bound:.3f}], "
                              f"count={count}, z_val={z_val:.3f}, alpha=beta={alpha:.2f}")

                        # Genera i punti con distribuzione beta
                        beta_samples = np.random.uniform(left_bound, right_bound, count)

                        # Trasforma nell'intervallo [left_bound, right_bound]
                        y_new = beta_samples

                        # Aggiungi ai risultati
                        xs_redistributed.extend([x_val] )
                        ys_redistributed.extend(y_new)
                        zs_redistributed.extend([z_val] )
                        ars_redistributed.extend([ars[indices[i]]])

    return (np.array(xs_redistributed), 
            np.array(ys_redistributed), 
            np.array(zs_redistributed),
            np.array(ars_redistributed))

def redistribute_xs_uniform(xs, ys, zs, ars):

    
    # Contiamo quante volte appare ogni valore unico di x
    unique_xs, counts = np.unique(xs, return_counts=True)
    
    print("Valori unici di xs e le loro frequenze:")
    for val, count in zip(unique_xs, counts):
        print(f"xs = {val}: {count} occorrenze")
    
    xs_redistributed = []
    ys_redistributed = []
    zs_redistributed = []
    as_redistributed = []
    # Per ogni valore unico di x
    for i, (x_val, count) in enumerate(zip(unique_xs, counts)):
        # Trova gli indici dove xs == x_val
        indices = np.where(xs == x_val)[0]
        
        # Definisci l'intervallo per questo valore
        if i == 0:
            # Primo intervallo: da x_val - 0.5 a x_val + 0.5
            x_next = unique_xs[i+1]
            #left_bound = (x_prev + x_val) / 2
            right_bound = (x_val + x_next) / 2
            left_bound = 2*x_val - right_bound 
        else:
            # Intervalli successivi: da (x_val + x_prev)/2 a (x_val + x_next)/2
            x_prev = unique_xs[i-1]
            if i < len(unique_xs) - 1:
                x_next = unique_xs[i+1]
                left_bound = (x_prev + x_val) / 2
                right_bound = (x_val + x_next) / 2
            else:
                # Ultimo intervallo
                left_bound = (x_prev + x_val) / 2
                right_bound = x_val + (x_val - x_prev) / 2
        
        # Genera valori uniformemente distribuiti nell'intervallo
        x_redistributed = np.linspace(left_bound, right_bound, count, endpoint=False)
        # Aggiungi un piccolo offset casuale per evitare sovrapposizioni perfette
        x_redistributed = np.random.uniform(left_bound, right_bound, count)
        
        # Aggiungi i valori redistribuiti
        xs_redistributed.extend(x_redistributed)
        ys_redistributed.extend(ys[indices])
        zs_redistributed.extend(zs[indices])
        as_redistributed.extend(ars[indices])
        
        print(f"Intervallo per xs={x_val}: [{left_bound:.3f}, {right_bound:.3f}] con {count} punti")
    
    return np.array(xs_redistributed), np.array(ys_redistributed), np.array(zs_redistributed), np.array(as_redistributed)


def redistribute_ys_multinomial(xs, ys, zs, ars, y_intervals=None):
    """
    Redistribuisce i valori ys secondo una distribuzione multinomiale
    basata sui valori zs (posterior_per_area).
    
    Parametri:
    - xs, ys, zs: array dei valori originali
    - y_intervals: lista degli intervalli y (se None, viene calcolata automaticamente)
    
    Ritorna:
    - xs, ys_redistributed, zs: array con ys redistribuiti
    """
    
    # Identifica i valori unici di y (centri degli intervalli)
    unique_ys = np.unique(ys)
    print(f"Valori unici di ys (centri intervalli): {unique_ys}")
    
    # Se non sono specificati gli intervalli, li calcoliamo
    if y_intervals is None:
        y_intervals = []
        for i, y_val in enumerate(unique_ys):
            if i == 0:
                if len(unique_ys) > 1:
                    y_next = unique_ys[i+1]
                    left_bound = y_val - (y_next - y_val) / 2
                    right_bound = (y_val + y_next) / 2
                else:
                    left_bound = y_val - 0.5
                    right_bound = y_val + 0.5
            elif i == len(unique_ys) - 1:
                y_prev = unique_ys[i-1]
                left_bound = (y_prev + y_val) / 2
                right_bound = y_val + (y_val - y_prev) / 2
            else:
                y_prev = unique_ys[i-1]
                y_next = unique_ys[i+1]
                left_bound = (y_prev + y_val) / 2
                right_bound = (y_val + y_next) / 2
            
            y_intervals.append((left_bound, right_bound))
    
    print("Intervalli y calcolati:")
    for i, (left, right) in enumerate(y_intervals):
        print(f"  y={unique_ys[i]}: [{left:.3f}, {right:.3f}]")
    
    ys_redistributed = []
    zs *= ars
    # Per ogni punto
    for i in range(len(xs)):
        y_val = ys[i]
        z_val = zs[i]  # posterior_per_area
        
        # Trova l'indice dell'intervallo corrispondente
        y_idx = np.where(unique_ys == y_val)[0][0]
        left_bound, right_bound = y_intervals[y_idx]
        
        # Usa z_val come peso per la distribuzione multinomiale
        # Più alto è z_val, più il punto dovrebbe essere vicino al centro
        # Usiamo una distribuzione beta per concentrare i punti
        
        # Normalizziamo z_val per usarlo come parametro della distribuzione
        # Valori alti di z_val -> concentrazione verso il centro
        # Valori bassi di z_val -> distribuzione più uniforme
        
        # Parametri per la distribuzione beta
        # Se z_val è alto, alpha e beta sono alti (concentrazione al centro)
        # Se z_val è basso, alpha e beta sono bassi (più uniforme)
        concentration = max(1.0, z_val * 10)  # Scala il peso
        alpha_param = concentration
        beta_param = concentration
        
        # Genera un valore dalla distribuzione beta e scalalo all'intervallo
        beta_sample = np.random.beta(alpha_param, beta_param)
        y_new = left_bound + beta_sample * (right_bound - left_bound)
        
        ys_redistributed.append(y_new)
    
    return xs, np.array(ys_redistributed), zs

class AdaptiveBivariatePolyaTreeMultiAlpha:
    def __init__(self, base_depth=4, max_depth=12, density_threshold=0.0001, min_mass=0.0001,   
                 alpha1=0.5, beta1=0.5, alpha2=0.5, beta2=0.5, gamma=2.0,   
                 domain=((0, 1), (0, 1)), kind=None):  
        self.base_depth = base_depth  
        self.max_depth = max_depth  
        self.density_threshold = density_threshold  
        self.min_mass = min_mass  
        # Parameters for x-direction Beta  
        self.alpha1 = alpha1  
        self.beta1 = beta1  
        # Parameters for y-direction Beta  
        self.alpha2 = alpha2  
        self.beta2 = beta2  
        self.gamma = gamma  
        self.domain = domain  
        self.counts = {}  
        self.counts[()] = 0  
        self.total_mass = 0
        self.kind=kind
    def _compute_beta_probability(self, count, total, alpha, beta, depth):  
        """Compute Beta posterior probability for a single direction"""  
        c = self.gamma  
        depth_factor = c ** depth  
        if self.kind == 'depth':
            #alpha = alpha * (1 +  depth / self.max_depth)
            alpha=1
        elif self.kind == 'density':
            alpha = alpha * (1 + beta * np.exp(-self.gamma * count/total))
        elif self.kind == 'curvature':
            alpha = alpha * (1 + 1.0 * abs(depth * count/total))
        elif self.kind == 'hybrid':
            alpha = alpha * (1 + beta * np.exp(-self.gamma * count/total))
        post_alpha = alpha * depth_factor + count/total  
        post_beta = alpha * depth_factor + (1 - count/total)  
        
        return post_alpha / (post_alpha + post_beta)
    
    
        
    def _compute_bivariate_density(self, node):  
        (x_left, x_right), (y_bottom, y_top) = self._get_node_bounds(node)  
        x_min, x_max = self.domain[0]  
        y_min, y_max = self.domain[1]  
          
        area_tot = (x_max - x_min) * (y_max - y_min)  
        area = (x_right - x_left) * (y_top - y_bottom)  
        count = self.counts.get(node, 0)  
        depth = len(node)  
          
        # Compute effective depth in each direction  
        depth_x = np.log2((x_max - x_min)/(x_right - x_left))  
        depth_y = np.log2((y_max - y_min)/(y_top - y_bottom))  
          
        # Compute probabilities for both directions  
        prob_x = self._compute_beta_probability(  
            count, self.total_mass, self.alpha1, self.beta1, depth_x)  
        prob_y = self._compute_beta_probability(  
            count, self.total_mass, self.alpha2, self.beta2, depth_y)
        joint_density = prob_x * prob_y  
          
        return joint_density if area > 0 else 0
    
    def _get_quadrant(self, x, y, x_mid, y_mid):
        if x <= x_mid:
            return 0 if y <= y_mid else 1
        else:
            return 2 if y <= y_mid else 3
        
    def _should_split(self, node, mass, input_pixel_size):  
        if len(node) >= self.max_depth:  
            return False  
              
        (x_left, x_right), (y_bottom, y_top) = self._get_node_bounds(node)  
        node_width = x_right - x_left  
        node_height = y_top - y_bottom  
          
        if node_width/2 < input_pixel_size[0] or node_height/2 < input_pixel_size[1]:  
            return False  
              
        if len(node) < self.base_depth:  
            return True  
              
        density = self._compute_bivariate_density(node)  
        return (mass >= self.min_mass * self.total_mass and   
                density >= self.density_threshold and   
                len(node) < self.max_depth)
    
    def _get_node_bounds(self, node):
        x_left, x_right = self.domain[0]
        y_bottom, y_top = self.domain[1]
        for q in node:
            x_mid = (x_left + x_right) / 2.0
            y_mid = (y_bottom + y_top) / 2.0
            if q in [0, 1]:
                x_right = x_mid
            else:
                x_left = x_mid
            if q in [0, 2]:
                y_top = y_mid
            else:
                y_bottom = y_mid
        return (x_left, x_right), (y_bottom, y_top)
    
    def _compute_local_density(self, node):
        (x_left, x_right), (y_bottom, y_top) = self._get_node_bounds(node)
        x_min, x_max = self.domain[0]
        y_min, y_max = self.domain[1]
        area_tot = (x_max - x_min)*(y_max-y_min)
        area = (x_right - x_left) * (y_top - y_bottom)
        count = self.counts.get(node, 0)
        coef_prob = count / (self.total_mass * area)
        depth = len(node)
        depth1 = np.log(area_tot/area)/np.log(4.0)
        c=3.0
        #post_alpha = self.alpha0 * (c ** depth1) + count/self.total_mass
        #post_beta = self.beta * (c ** depth1) + (1 - count/self.total_mass) 
        
        # Compute posterior mean
        #posterior_mean = post_alpha / (post_alpha + post_beta)
        
        return count / (self.total_mass * area) if area > 0 else 0
        #return posterior_mean if area > 0 else 0
    
    def _should_split1(self, node, mass, input_pixel_size):
        if len(node) >= self.max_depth:
            return False
            
        # Get current node dimensions
        (x_left, x_right), (y_bottom, y_top) = self._get_node_bounds(node)
        node_width = x_right - x_left
        node_height = y_top - y_bottom
        
        # Check if splitting would create cells smaller than input pixels
        if node_width/2 < input_pixel_size[0] or node_height/2 < input_pixel_size[1]:
            return False
            
        if len(node) < self.base_depth:
            return True
            
        density = self._compute_local_density(node)
        return (mass >= self.min_mass * self.total_mass and 
                density >= self.density_threshold and 
                len(node) < self.max_depth)
    
    def _update_counts_from_histogram(self, hist_2d, input_pixel_size):
        """Update counts using 2D histogram data"""
        self.total_mass = np.sum(hist_2d)
        
        # Get histogram dimensions
        ny, nx = hist_2d.shape
        dx = (self.domain[0][1] - self.domain[0][0]) / nx
        dy = (self.domain[1][1] - self.domain[1][0]) / ny
        
        points_to_process = [((), hist_2d)]
        while points_to_process:
            node, node_hist = points_to_process.pop(0)
            
            # Compute mass in current node
            self.counts[node] = np.sum(node_hist)
            
            if self._should_split(node, self.counts[node], input_pixel_size):
                # Split histogram into quadrants
                ny, nx = node_hist.shape
                mid_y, mid_x = ny//2, nx//2
                
                quadrants = {
                    0: node_hist[:mid_y, :mid_x],
                    1: node_hist[mid_y:, :mid_x],
                    2: node_hist[:mid_y, mid_x:],
                    3: node_hist[mid_y:, mid_x:]
                }
                
                for q in range(4):
                    if np.sum(quadrants[q]) > 0:  # Only process non-empty quadrants
                        points_to_process.append((node + (q,), quadrants[q]))
    
    def fit(self, hist_2d):
        """Fit the Polya tree using a 2D histogram"""
        # Calculate input pixel size
        print(hist_2d.shape)
        ny, nx = hist_2d.shape
        dx = (self.domain[0][1] - self.domain[0][0]) / nx
        dy = (self.domain[1][1] - self.domain[1][0]) / ny
        input_pixel_size = (dx, dy)
        
        self._update_counts_from_histogram(hist_2d, input_pixel_size)
        return self

# Test the modified class with a simple example
if __name__ == "__main__":
    # Create a simple test histogram
    nx, ny = 50, 50
    x = np.linspace(0, 1, nx)
    y = np.linspace(0, 1, ny)
    X, Y = np.meshgrid(x, y)
    
    # Create a simple gaussian bump
    hist_2d = np.exp(-((X-0.5)**2 + (Y-0.5)**2)/0.1)
    
    # Initialize and fit the tree
    tree = AdaptiveBivariatePolyaTreeMultiAlpha(
        base_depth=2,
        max_depth=8,
        density_threshold=0.01,
        min_mass=0.01
    )
    tree.fit(hist_2d)
    
    # Print some basic information
    print("Total mass:", tree.total_mass)
    print("Number of nodes:", len(tree.counts))
    
    # Plot the input histogram
    import matplotlib.pyplot as plt
    
    plt.figure(figsize=(10, 4))
    
    plt.subplot(121)
    plt.imshow(hist_2d, origin='lower', extent=[0, 1, 0, 1])
    plt.colorbar(label='Density')
    plt.title('Input Histogram')
    
    # Plot the node structure
    plt.subplot(122)
    for node in tree.counts:
        if node:  # Skip root node
            (x_left, x_right), (y_bottom, y_top) = tree._get_node_bounds(node)
            plt.plot([x_left, x_right], [y_bottom, y_bottom], 'k-', alpha=0.5)
            plt.plot([x_left, x_right], [y_top, y_top], 'k-', alpha=0.5)
            plt.plot([x_left, x_left], [y_bottom, y_top], 'k-', alpha=0.5)
            plt.plot([x_right, x_right], [y_bottom, y_top], 'k-', alpha=0.5)
    
    plt.xlim(0, 1)
    plt.ylim(0, 1)
    plt.title('Tree Structure')
    
    plt.tight_layout()
    plt.show()

def generate_chirp_data(n_points=200, fake=True):
    if fake:
        # Generate clean theoretical curve first
        t_clean = np.linspace(0, 1, n_points)
        f_clean = 20 + 980 * t_clean**2

        # Generate noisy data
        t = t_clean + 0.02 * np.random.randn(n_points)
        f = f_clean + 20 * np.random.randn(n_points)

        # Normalize both clean and noisy data
        t = (t - t.min()) / (t.max() - t.min())
        f = (f - f.min()) / (f.max() - f.min())
        t_clean = (t_clean - t_clean.min()) / (t_clean.max() - t_clean.min())
        f_clean = (f_clean - f_clean.min()) / (f_clean.max() - f_clean.min())

        return np.column_stack([t, f]), np.column_stack([t_clean, f_clean])
    
        

class AdaptiveBivariatePolyaTreeMultiAlphaExtended(AdaptiveBivariatePolyaTreeMultiAlpha):
    def compute_posteriors_alphas(self, resolution=(50,50)):
        # Create grid centers using resolution
        nx, ny = resolution
        x = np.linspace(self.domain[0][0], self.domain[0][1], nx)
        y = np.linspace(self.domain[1][0], self.domain[1][1], ny)
        X, Y = np.meshgrid(x, y)
        
        densities = np.zeros_like(X)
        alpha_maps = {
            'depth': np.zeros_like(X),
            'density': np.zeros_like(X),
            'curvature': np.zeros_like(X),
            'hybrid': np.zeros_like(X)
        }
        normalized_alpha_maps = {
            'depth': np.zeros_like(X),
            'density': np.zeros_like(X),
            'curvature': np.zeros_like(X),
            'hybrid': np.zeros_like(X)
        }
        pixel_areas = np.zeros_like(X)
        
        for i in range(X.shape[0]):
            for j in range(X.shape[1]):
                x_pt, y_pt = X[i,j], Y[i,j]
                node = ()
                prob = 1.0
                while True:
                    (x_left, x_right), (y_bottom, y_top) = self._get_node_bounds(node)
                    area = (x_right - x_left) * (y_top - y_bottom)
                    pixel_areas[i,j] = area
                    
                    local_density = self._compute_local_density(node)
                    depth = len(node)
                    
                    # Compute different alpha definitions
                    alpha_depth = self.alpha0 * (1 + self.lam * depth / self.max_depth)
                    alpha_density = self.alpha0 * (1 + self.beta * np.exp(-self.gamma * local_density))
                    alpha_curvature = self.alpha0 * (1 + self.eta * abs(depth * local_density))
                    alpha_hybrid = alpha_depth * (1 + self.beta * np.exp(-self.gamma * local_density))
                    
                    # Store the alpha values
                    alpha_maps['depth'][i,j] = alpha_depth
                    alpha_maps['density'][i,j] = alpha_density
                    alpha_maps['curvature'][i,j] = alpha_curvature
                    alpha_maps['hybrid'][i,j] = alpha_hybrid
                    
                    normalized_alpha_maps['depth'][i,j] = alpha_depth / area if area>0 else 0
                    normalized_alpha_maps['density'][i,j] = alpha_density / area if area>0 else 0
                    normalized_alpha_maps['curvature'][i,j] = alpha_curvature / area if area>0 else 0
                    normalized_alpha_maps['hybrid'][i,j] = alpha_hybrid / area if area>0 else 0
                    
                    # Decide to split or not
                    mass = self.counts.get(node, 0)
                    input_dx = (self.domain[0][1] - self.domain[0][0]) / resolution[0]
                    input_dy = (self.domain[1][1] - self.domain[1][0]) / resolution[1]
                    input_pixel_size = (input_dx, input_dy)
                    
                    if not self._should_split(node, mass, input_pixel_size):
                        densities[i,j] = prob * local_density
                        break
                    
                    x_mid = (x_left + x_right)/2.0
                    y_mid = (y_bottom + y_top)/2.0
                    quadrant = self._get_quadrant(x_pt, y_pt, x_mid, y_mid)
                    node = node + (quadrant,)
                    prob /= 4.0

        return X, Y, densities, alpha_maps, normalized_alpha_maps, pixel_areas
tree_ext = AdaptiveBivariatePolyaTreeMultiAlphaExtended(
        base_depth=1,
        max_depth=8,
        density_threshold=0.00001,
        min_mass=0.00001,
        kind='depth'
    )

map_time = np.argmax(q.value)
row, col = np.unravel_index(map_time, q.value.shape)
print(row, col)
qvalue = q.value[row-200: row+10, :]/(np.sum(q.value))
tree_ext.fit(np.abs(qvalue.T))
```
