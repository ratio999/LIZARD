```python
import numpy as np
import matplotlib.pyplot as plt
import pywt
from scipy.fft import fft, ifft, fftfreq, fftshift

# 1. Parametri e segnale sine-Gaussian
fs = 4096  # frequenza di campionamento
t = np.linspace(-0.5, 0.5, fs, endpoint=False)
f0 = 150  # frequenza centrale
sigma = 0.05

# Segnale: sine-Gaussian
s = np.exp(-t**2 / (2*sigma**2)) * np.sin(2*np.pi*f0*t)

plt.figure(figsize=(10, 3))
plt.plot(t, s)
plt.title("Segnale Sine-Gaussian")
plt.xlabel("Tempo (s)")
plt.grid(True)
plt.show()

# 2. Trasformata Wavelet standard
wavelet = 'db6'
coeffs = pywt.wavedec(s, wavelet, level=6)
reconstructed_wavelet = pywt.waverec(coeffs, wavelet)

# Sparsità: numero di coefficienti sopra soglia
threshold = 0.01
sparse_wavelet = sum([np.sum(np.abs(c) > threshold) for c in coeffs])

# 3. Trasformata di Fresnel
def fresnel_transform(signal):
    N = len(signal)
    x = np.arange(N) - N//2
    phase = np.exp(1j * np.pi * (x**2) / N)
    return ifft(fft(signal) * phase)

s_fresnel = fresnel_transform(s)
def inverse_fresnel_transform(signal):
    N = len(signal)
    x = np.arange(N) - N//2
    phase = np.exp(-1j * np.pi * (x**2) / N)
    return np.real(ifft(fft(signal) * phase))

# 4. Composizione Fresnelet (Fresnel + Wavelet)
fresnelet_coeffs = pywt.wavedec(s_fresnel, wavelet, level=6)
reconstructed_fresnelet = inverse_fresnel_transform(pywt.waverec(fresnelet_coeffs, wavelet))
#reconstructed_fresnelet = pywt.waverec(fresnelet_coeffs, wavelet)

# Sparsità fresnelet
sparse_fresnelet = sum([np.sum(np.abs(c) > threshold) for c in fresnelet_coeffs])

# 5. Confronto visivo

plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(t, reconstructed_wavelet, label='Wavelet')
plt.plot(t, s, '--', alpha=0.5, label='Originale')
plt.title("Ricostruzione Wavelet")
plt.legend()

plt.subplot(2, 1, 2)
plt.plot(t, reconstructed_fresnelet, label='Fresnelet', color='orange')
plt.plot(t, s, '--', alpha=0.5, label='Originale')
plt.title("Ricostruzione Fresnelet")
plt.legend()
plt.tight_layout()
plt.show()

# 6. Sparsità comparata
print(f"Sparsità wavelet: {sparse_wavelet} coefficienti sopra soglia")
print(f"Sparsità fresnelet: {sparse_fresnelet} coefficienti sopra soglia")
```

```python
def create_tf_dictionary(T, F, n_gaussians=8):
    dict_matrix = []
    n_points = T.size
    
    # Constant term
    dict_matrix.append(np.ones(n_points))
    
    # Linear terms
    dict_matrix.append(T.flatten())
    dict_matrix.append(F.flatten())
    
    # Quadratic and cross terms
    dict_matrix.append(T.flatten()**2)
    dict_matrix.append(F.flatten()**2)
    dict_matrix.append(T.flatten() * F.flatten())
    
    # Create Gaussian bumps distributed over the domain
    t_min, t_max = np.min(T), np.max(T)
    f_min, f_max = np.min(F), np.max(F)
    t_centers = np.linspace(t_min, t_max, n_gaussians)
    f_centers = np.linspace(f_min, f_max, n_gaussians)

    # Define widths
    sigma_t = (t_max - t_min) / (2 * n_gaussians)
    sigma_f = (f_max - f_min) / (2 * n_gaussians)

    for t0 in t_centers:
        for f0 in f_centers:
            gaussian = np.exp(-((T.flatten()-t0)**2)/(2*sigma_t**2) - ((F.flatten()-f0)**2)/(2*sigma_f**2))
            dict_matrix.append(gaussian)
    
    return np.column_stack(dict_matrix)
```

```python
X_dict = create_tf_dictionary(T, F)
print('Dictionary shape:', X_dict.shape)

# Flatten the intensity data
y = qdata.flatten()

# Fit sparse regression
n_terms = 20
print('Fitting OMP...')
omp = OrthogonalMatchingPursuit(n_nonzero_coefs=n_terms)
omp.fit(X_dict, y)

# Reconstruct the signal
qdata_pred = omp.predict(X_dict).reshape(qdata.shape)
qdata =qdata[:-1,:-1]  
#qdata =qdata[:-1,:-1]  
qdata_pred =qdata_pred[:-1,:-1]  
# Plot results
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20, 5))
times_mesh=times
freqs_mesh=freqs
#times_mesh=np.pad(times, (1,0), times[0]-1)
#freqs_mesh=np.pad(freqs, (1,0), times[0]-1)
print(times_mesh.shape, freqs_mesh.shape, qdata.shape)
# Original
#times_mesh=times[:-1]
#freqs_mesh=freqs[:-1]

im1 = ax1.pcolormesh(times_mesh - times_mesh[0], freqs_mesh, qdata.T, shading='flat', cmap='viridis')  
ax1.set_title('Original Q-transform')
ax1.set_xlabel('Time (s) from start')
ax1.set_ylabel('Frequency (Hz)')
plt.colorbar(im1, ax=ax1)

# Reconstruction
im2 = ax2.pcolormesh(times_mesh - times_mesh[0], freqs_mesh, qdata_pred.T, 
                     shading='flat', cmap='viridis')
ax2.set_title('OMP Reconstruction')
ax2.set_xlabel('Time (s) from start')
ax2.set_ylabel('Frequency (Hz)')
plt.colorbar(im2, ax=ax2)

# Residuals
im3 = ax3.pcolormesh(times_mesh - times_mesh[0], freqs_mesh, qdata.T - qdata_pred.T, 
                     shading='flat', cmap='viridis')
ax3.set_title('Residuals')
ax3.set_xlabel('Time (s) from start')
ax3.set_ylabel('Frequency (Hz)')
plt.colorbar(im3, ax=ax3)

plt.tight_layout()
plt.show()

# Print metrics
r2_val = r2_score(qdata.flatten(), qdata_pred.flatten())
print('R² score:', r2_val)
print('Number of non-zero coefficients:', np.sum(np.abs(omp.coef_) > 1e-10))
```
