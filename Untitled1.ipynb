```python

times=hdata.times
times1 = hdata1.times
print(type(times)) 
signal_gw = hdata
#print(type(signal_gw.real))

noise_level = 1*np.std(signal_gw.data)


signal_noisy = hdata1

# Pure noise with same standard deviation as signal
signal_noise = np.random.normal(0, np.std(signal_noisy), signal_noisy.shape)
signal_noise = TimeSeries(signal_noise, times=times)
# Define Takens embedding function
def takens_embedding(time_series, embedding_dim, delay):
    white=time_series#.whiten(4,2)
    #white=white.value(gpstimes)
    timeseries=white.data
    n = len(time_series)
    M = n - (embedding_dim - 1) * delay
    if M <= 0:
        raise ValueError('Time series is too short')
    embedded = np.empty((M, embedding_dim))
    for i in range(M):
        embedded[i] = timeseries[i:(i + embedding_dim*delay):delay]
    return embedded

# Parameters for Takens embedding
embedding_dim = 3
delay = 20

# Compute embeddings
embedded_gw = takens_embedding(signal_gw, embedding_dim, delay)
embedded_noisy = takens_embedding(signal_noisy, embedding_dim, delay)
embedded_noise = takens_embedding(signal_noise, embedding_dim, delay)

# Compute spectrograms
from gwpy.timeseries import TimeSeries
def compute_spectrogram(signal):
    q= signal.q_transform(frange=(30, 500))

# Create the plot for the Q-transform
    plot = q.plot()
    plot.colorbar(label="Normalised energy")
    return q 

q1 = compute_spectrogram(signal_gw)
q2 = compute_spectrogram(signal_noisy)
q3 = compute_spectrogram(signal_noise)

# Create the plot
fig = plt.figure(figsize=(16, 14))

row_titles = ['GW178017', 'Noisy Chirp', 'Pure Noise']
col_titles = ['Time Domain', 'Spectrogram', "Takens' Embedding"]

################ Row 1: GW Chirp ################
# Time Domain
ax1 = plt.subplot(3, 2, 1)
ax1.plot(times, np.array(signal_gw.whiten(2,1).data), color='b')
ax1.set_title(col_titles[0])
ax1.set_ylabel(row_titles[0])
ax1.set_xlabel('Time (s)')

# Spectrogram

# Takens Embedding
ax3 = plt.subplot(3, 2, 2, projection='3d')
ax3.plot(embedded_gw[:,0], embedded_gw[:,1], embedded_gw[:,2], lw=0.8, color='b')
ax3.set_title(col_titles[2])
ax3.set_xlabel('x(t)')
ax3.set_ylabel('x(t+delay)')
ax3.set_zlabel('x(t+2*delay)')

################ Row 2: Noisy Chirp ################
# Time Domain
ax4 = plt.subplot(3, 2, 3)
ax4.plot(times, np.array(signal_noisy.data), color='r')
ax4.set_ylabel(row_titles[1])
ax4.set_xlabel('Time (s)')

# Takens Embedding
ax6 = plt.subplot(3, 2, 4, projection='3d')
ax6.plot(embedded_noisy[:,0], embedded_noisy[:,1], embedded_noisy[:,2], lw=0.8, color='r')
ax6.set_xlabel('x(t)')
ax6.set_ylabel('x(t+delay)')
ax6.set_zlabel('x(t+2*delay)')

################ Row 3: Pure Noise ################
# Time Domain
ax7 = plt.subplot(3, 2, 5)
ax7.plot(times, np.array(signal_noise.data), color='g')
ax7.set_ylabel(row_titles[2])
ax7.set_xlabel('Time (s)')

# Spectrogram


# Takens Embedding
ax9 = plt.subplot(3, 2, 6, projection='3d')
ax9.plot(embedded_noise[:,0], embedded_noise[:,1], embedded_noise[:,2], lw=0.8, color='g')
ax9.set_xlabel('x(t)')
ax9.set_ylabel('x(t+delay)')
ax9.set_zlabel('x(t+2*delay)')

plt.tight_layout()
plt.show()

print('Generated plots using IMRPhenomXAS waveform for a 30-30 solar mass binary black hole merger at 400 Mpc.')
```
